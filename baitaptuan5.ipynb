```python
#VIết hàm kiểm tra một số có phải số hoàn hảo hay không, xử lý trên dữ liệu kiểu số nguyên.
def sohoanhao(a):
    tong_uoc = 0
    for i in range(1,a):
       if a % i == 0:
          tong_uoc += i
    if tong_uoc == a:
      return(a,'là số hoàn hảo')
    else:
      return(a,"là số không hoàn hảo ")
try:
    
   b = int(input())
   print(*sohoanhao(b))
except ValueError:
    print('hayx nhập số Nguyên')
```

```python
#Viết hàm trả về tổng các chữ số của số nguyên dương n
def tong(a):
    tong = 0
    while a > 0:
        tong += a % 10
        a = a // 10
    return tong
n = int(input())
print(tong(n))
```

```python
#Viết hàm tính giai thừa. Nhập vào số nguyên n, trả về giai thừa của n (n!)
def giaithua(a):
    tich = 1
    i = 1
    while i <= a:
        tich *= i
        i +=1
    return(tich)
try:
    n = int(input())
    print('n!=',giaithua(n))
except ValueError:
    print('vui lòng nhập số hợp lệ')
```

```python
import math
def sontmax(n):
    res = 1
    for i in range(2, int(math.sqrt(n))+1):
        if n % i == 0:
            res = i
            
            while n % i == 0:
                
                n = n // i
    if n > 1:
        res = n
    return res    
a = int(input())
print(sontmax(a))

```

```python
def maytinh(so1,so2,pheptoan):
    if pheptoan == "+":
        return so1 + so2
    elif pheptoan == "-":
        return so1 - so2
    elif pheptoan == "*":
        return so1 * so2
    elif pheptoan == "/":
        if so2 == 0:
            return "ko thể chia cho 0"
        return so1 / so2
    else:
        print('toán tử không hợp lệ')
try:        
 num1str,pheptinh,num2str = input().split()
 num1 = float(num1str)
 num2 = float(num2str)
 print('kết quả của phép toán trên là',round(maytinh(num1,num2,pheptinh),2))
except ValueError:
    print('vui lòng nhập số thực và phép toán phù hợp')
except Exception:
    print('Đã xảy ra lỗi không mong muốn')
```
