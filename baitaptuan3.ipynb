```python
a = int(input())
b = int(input())
c = int(input())

if a + b <= c or a + c <= b or b + c <= a:
    print("Không phải tam giác")
elif a == b == c:
    print("Tam giác đều")
elif a == b or a == c or b == c:
    print("Tam giác cân")
else:
    print("Tam giác thường")
```
