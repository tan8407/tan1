```python
#bài 4
h = int(input())
for i in range(1,h+1):
    print(" "*(h-i), end="")
    if i == 1:
        print("*")
    elif i == h:
        print("*" * (2*h-1))
    else:
        print ("*" + " " * (2*i-3) + "*")
```

```python
# bài 13
maxx = None   # maxx là số nhập lớn nhất
minn = None   #minn_____________nhỏ nhất
print("bắt đầu nhập số nguyên n (nhập -1 để dựng lại)")
for i in range(1000):
    n = int(input("nhập số: "))
    if n == -1:
        break
    if minn is None:
        maxx = n
        minn = n
    else:
       if maxx < n:
          maxx = n
       if minn > n:
          minn =n
if n == -1:
    print("bạn chưa nhập số nào ngoài -1")
else:
    print("số lớn nhất trong các số vừa nhập", maxx)
    print("số nhỏ nhất trong các số vừa nhập", minn)
        
```

```python
#bài 15
try:
   n = int(input("nhập vào môtj số nguyên dương"))
   so_chan = 0
   so_le = 0
   so_chu_so = 0
   while n > 0:
      n = n // 10
      so_chu_so = n % 10
      if so_chu_so % 2 == 0:
        so_chan += 1
      else:

        so_le += 1
   print("số chữ số lẻ là", so_le)
   print("số chữ số chẵn là", so_chan)
except ValueError :
   print("vui lòng nhập số hợp lệ")
```

```python
#bài 20
try:
    n = int(input("nhập vào môtj số nguyên dương"))
    if n < 0:
        print("nhập số dương")
    chu_so = 0
    cac_so_nguyen_to = 0
    while n > 0:
        chu_so = n % 10
        if chu_so == 1 or chu_so == 2 or chu_so == 3 or chu_so == 5 or chu_so == 7:
           cac_so_nguyen_to += 1
        n = n // 10
    print(cac_so_nguyen_to)
except ValueError :
    print("nhập số nguyên")
           
```
