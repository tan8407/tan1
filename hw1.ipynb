```python
a = float(input())
b = float(input())

r = min(a, b) / 2
S = a * b - 3.14 * r * r

print(f"{S:.2f}")

```

```python
c = input()

if 'A' <= c <= 'Z':
    print(chr(ord(c) + 32))
elif 'a' <= c <= 'z':
    print(chr(ord(c) - 32))
```

```python
import math

a = float(input())
b = float(input())
c = float(input())

if a + b > c and a + c > b and b + c > a:
    p = (a + b + c) / 2
    S = math.sqrt(p * (p - a) * (p - b) * (p - c))
    print(f"{S:.1f}")
else:
    print("Khong phai 3 canh cua tam giac")
```
