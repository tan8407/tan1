```python
n, m = map(int, input().split())
a = [list(map(int, input().split())) for _ in range(n)]

for row in a:
    for x in row:
        print(f"{x:4}", end="")
    print()
```

```python
def max_index(lst):
    return lst.index(max(lst))
a = list(map(int, input().split()))
print(max_index(a))
```

```python
def find_k(lst, k):
    return lst.index(k) if k in lst else -1
a = list(map(int, input().split()))
k = int(input())
print(find_k(a,k))
```

```python
words = input().split()
d = {}

for i, w in enumerate(words):
    d.setdefault(w, []).append(i)

for k in d:
    d[k] = tuple(d[k])

print(d)
```

```python
def move_zero_front(lst):
    zeros = [i for i in lst if i == 0]
    others = [i for i in lst if i != 0]
    return zeros + others
a = list(map(int, input().split()))
print(move_zero_front(a))
```

```python
def local_max_sum(lst):
    s = 0
    for i in range(1, len(lst)-1):
        if lst[i] > lst[i-1] and lst[i] > lst[i+1]:
            s += lst[i]
    return s
a = list(map(int, input().split()))
print(local_max_sum(a))
```
