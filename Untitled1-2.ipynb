```python
#bài 3
import math
a,b,c =map(float,input().split())
chu_vi = a+b+c
p = chu_vi/2
dien_tich = math.sqrt(p*(p-a)*(p-b)*(p-c))
#p là công thức tính diện tichs
if a<=0 or b<=0 or c<=0 or a+b<=c or a+c<=b or b+c<=a :
    print("không là ba cạnh của tam giác")
else:
    print("diện tích tam giác là",dien_tich)
    print("chu vi tam giác là",chu_vi)
#vế kiểm tra tam giác    
if a==b==c :
    print("đây là tam giác vuông")
elif a==b or b==c or c==a :
    print("đây là tam giác đều")
elif a**2+b**2==c**2 or b**2+c**2==a**2 or a**2+c**2==b**2:
    print("đây là tam giác vuông")
elif a<=0 or b<=0 or c<=0 or a+b<=c or a+c<=b or b+c<=a :
    print("      ")
else:
    print("đây là tam giác thường")
```

```python
#bài 5
import math
a,b,c = map(float,input().split())
# phương trình ax^2+bx+c
A = b**2-4*a*c
if A>0:
    X1 = (-b+math.sqrt(A))/(2*a)
    X2 = (-b-math.sqrt(A))/(2*a)
    print("phương trình có 2 nghiệm phân biệt", X1,"và",X2)
elif A==0:
    X = -b/2*a
    print("phương trình có nghiệm kép",X)
else:
    print("phương trình vô nghiệm")


```

```python
#bài 9
tong_giay = int(input("Nhập vào tổng số giây: "))
so_gio = tong_giay // 3600
so_giay_con_lai = tong_giay % 3600
so_phut = so_giay_con_lai // 60
so_giay = so_giay_con_lai % 60
print(f"Tổng thời gian là: {so_gio} giờ, {so_phut} phút, {so_giay} giây.")
```

```python
#bài 10
import math
xM,yM = map(float,input("toạ độ tâm M là: ").split())
xA,yA = map(float,input("toạ độ tâm A là: ").split())
r = float(input("bán kính đường tròn là: ")) 
khoang_cach_AM = math.sqrt((xM-xA)**2+(yM-yA)**2)
if khoang_cach_AM <= r:
    print("A thuộc đường tròn tâm M")
else:
    print("A không thuộc đường tròn tâm M")
```
