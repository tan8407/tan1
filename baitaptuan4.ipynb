```python
#btvn
h = int(input("nhập vào chiều cao tam giác"))
i = 1
while i <= h:
    
    print(" "*(h-i), end="")
    if i == 1:
        print("*")
    elif i == h :
        print("*"*(2*h-1))
    else:
        print("*"+" "*(i*2-3)+"*")
    i += 1
```

```python
#Nhập vào số nguyên dương a, kiểm tra xem a có phải là số hoàn hảo hay không.
#Số hoàn hảo là số có tổng các ước (ước nhỏ hơn nó) bằng chính nó, VD: số 6, 28…
n = int(input())
i = 1
tong_uoc = 0
while i < n:
    
    if n % i == 0:
        tong_uoc += i
    i += 1

if tong_uoc == n:
    print("n là số hoàn hảo")
else:
    print("n ko phải số hoàn hảo")
    
```

```python
#Nhập vào một dãy số nguyên, dừng nhập khi người dùng nhập -1. Sau khi nhập
#xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập. KHÔNG ĐƯỢC dùng mảng
maxx = None   # maxx là số nhập lớn nhất
minn = None   #minn_____________nhỏ nhất
print("bắt đầu nhập số nguyên n (nhập -1 để dựng lại)")
while True :
    n = int(input("nhập số: "))
    if n == -1:
       break
    if minn is None:
        maxx = n
        minn = n
    else:
       if maxx < n:
          maxx = n
       if minn > n:
          minn =n
if maxx is None:
    print("bạn chưa nhập số nào ngoài -1")
else:
    print("số lớn nhất trong các số vừa nhập", maxx)
    print("số nhỏ nhất trong các số vừa nhập", minn)
```

```python
#Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số chẵn, bao nhiêu
#chữ số lẻ
try:
   n = int(input("nhập vào môtj số nguyên dương"))
   so_chan = 0
   so_le = 0
   so_chu_so = 0
   if n <= 0:
       print("vui lòng nhập số nguyên dương")
   
   while n > 0:
      
      
      so_chu_so = n % 10
      if so_chu_so % 2 == 0:
        so_chan += 1
      else:

        so_le += 1
      n = n // 10
   print("số chữ số lẻ là", so_le)
   print("số chữ số chẵn là", so_chan)
except ValueError :
   print("vui lòng nhập số hợp lệ")
```

```python
#hập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số là số nguyên tố.
try:
    n = int(input("nhập vào môtj số nguyên dương"))
    if n < 0:
        print("nhập số dương")
    chu_so = 0
    cac_so_nguyen_to = 0
    while n > 0:
        chu_so = n % 10
        if chu_so == 1 or chu_so == 2 or chu_so == 3 or chu_so == 5 or chu_so == 7:
           cac_so_nguyen_to += 1
        n = n // 10
    print("n có",cac_so_nguyen_to)
except ValueError :
    print("nhập số nguyên")
```
