```python
# Nhập một dòng
s = input().split()

result = {}

# Tách các cặp key:value
for i in s:#duyệt qua từng cặp phân biệt bởi dấu cách
    key, value = i.split(":")#cắt nó tại dấu :
# nếu key đầu tiên xuất hiện thì thêm một thì tạo một cặp mới
    if key not in result:
        result[key] = [value]
    else:#có rồi thì thêm vào danh sách
      result[key].append(value)

print(result)
```

```python
#A5
a = list(map(int,input().split()))
#tạo một dicsionary
#key là các tên biến value là các số đếm khởi tạo từ 0
dic__ = { 'positives':0,
         "negatives":0,
         "zero":0}
for i in a:
    if i > 0:
        dic__['positives'] += 1
    elif i < 0:
        dic__['negatives'] += 1
    else:
        dic__['zero'] += 1
print(dic__)        
```

```python
a1 = input().split()
a2 = input().split()
dictt = {}
for i in a1: 
    k,v = i.split(':') #cắt các phantu của a1 và gán k,v với key và value của djctionary
    dictt[k] = int(v) #vì v là kí tự nhập từ bàn phím nên ép kiểu nó về dạng số nguyên
  #lưu nó vào thư viện dictt(k là key,v thành value của k đấy)
for i in a2:
    k,v = i.split(':')
    v = int(v)
    if k in dictt:  #nếu k đã có trong thư viện dictt
        dictt[k] += v  #thì vulue của k đó được cộng dồn
    else:
        dictt[k] = v #không thì in ra khoá k ứng với v là value tương ứng giống dòng for đầu tiên
for k in dictt: #duyệt qua từng cặp đã lưu ở các phép toán bên trên
#nếu không duyệt thì chỉ in ra k,v được nhận cuối cùng
    print(f"{k}:{dictt[k]}", end=" ")
        
```

```python
a = list(map(int,input().split()))
k = int(input())
danhsach =[]
n = len(a)
for i in range(n):
    for j in range(i+1,n):
        if a[i] + a[j] == k:
            sapxep = sorted((a[i],a[j]))  #sắp xếp 2 số trong cặp
            if sapxep not in danhsach:  #trách in ra các cặp trùng lặp
                danhsach.append(sapxep)
            
danhsach.sort()  # dùng để sắp xếp các số đầu tiên theo thứ tự của các cặp
print(danhsach)
            
```

```python
a = list(map(int,input().split()))
new_dict = {}
n = set(a)
for i in n:
    new_dict[i] = 0
    for j in a:
      if j == i:
        new_dict[i] += 1
maxxx = max(new_dict.values())
Kmax = [k for k, v in new_dict.items() if v == maxxx]
print(*Kmax)
    
```
