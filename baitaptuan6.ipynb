```python
def binary_search(arr,left,right,target):
    
    while left <= right:     #lặp lại đến khi 2 vế xét = nhau
        mid = (right + left)//2  
        if arr[mid] == target:
            return mid
        elif arr[mid] > target:
             right = mid -1
        else:
         left = mid + 1
    return -1
a = list(map(int,input('nhập vào list số nguyên tăng dần').split()))
b = int(input())
c = 0
d = len(a)-1
print(binary_search(a,c,d,b))


```

```python
def count_occurrences(arr, x) -> int:
    demso = 0
    for i in arr:
        if i == x:
            demso += 1
    return demso
a = list(map(int,input().split()))
k = int(input())
print(count_occurrences(a,k))

            
```

```python
def sonhieunhat(a):
    new_dict = {}
    n = set(a)
    for i in n:
       new_dict[i] = 0
       for j in a:
          if j == i:
            new_dict[i] += 1
    maxxx = max(new_dict.values())  # tìm ra max của value
    Kmax = [k for k, v in new_dict.items() if v == maxxx]  #k thuộc 1 key:value của new_dict tìm được k ứng với v bằng maxxx
    return(*Kmax,'xuất hiện nhiều nhất, sớm nhất,',maxxx,'lần')
    
a = list(map(int,input().split()))
print(*sonhieunhat(a))
```

```python
#cách 2
def sonhieunhat(a):
    new_dict = {}
    # Bước 1: Đếm số lần xuất hiện
    # (Dùng cách cũ của bạn hoặc cách này đều được)
    for i in a:
        if i in new_dict:
            new_dict[i] += 1
        else:
            new_dict[i] = 1
            
    maxxx = max(new_dict.values())

    # Bước 2: Tìm số "Sớm nhất"
    # Duyệt danh sách a gốc để đảm bảo thứ tự
    for k in a: 
        if new_dict[k] == maxxx:
            # Trả về ngay khi gặp số thỏa mãn đầu tiên
            return (k, 'xuất hiện nhiều nhất, sớm nhất,', maxxx, 'lần')

a = list(map(int, input().split()))
# Dùng dấu * để in đẹp không ngoặc
print(*sonhieunhat(a))
```

```python
def run_length_encoding(s):
    if not s:
        return ""

    result = ""
    count = 1

    for i in range(1, len(s)):
        if s[i] == s[i - 1]:
            count += 1
           
        else:
            result += str(count) + s[i]
            count = 1

    # thêm nhóm cuối cùng
    

    return result
f = input()
print(run_length_encoding(f))
```

```python
def most_frequent_smallest(arr):
    if not arr:
        return None   # hoặc xử lý theo yêu cầu đề

    freq = {}
    for x in arr:
        freq[x] = freq.get(x, 0) + 1

    max_freq = max(freq.values())

    # Lấy số nhỏ nhất trong các số có tần suất = max_freq
    result = min(x for x in freq if freq[x] == max_freq)

    return result
```

```python
def second_largest_and_second_smallest(s):
    if len(s)<2:
        raise ValueError("Expected ValueError for fewer than 2 distinct values")
    elif len(s)==2:
        return (max(s),min(s))
    else:
        s=sorted(set(s))
        return(s[-2],s[1])
s = list(map(int,input().split()))
print(second_largest_and_second_smallest(s))
```
